```python
# Check Docker installation
import subprocess

try:
    result = subprocess.run(["docker","--version"], capture_output=True, text=True, check=True)
    print("Docker is installed:", result.stdout.strip())
except subprocess.CalledProcessError:
    print("Docker is not installed or not found in PATH.")
```

```python
# Check Docker Compose

try:
    result = subprocess.run(["docker-compose","--version"], capture_output=True, text=True, check=True)
    print("Docker Compose is installed:", result.stdout.strip())
except subprocess.CalledProcessError:
    print("Docker Compose is not installed or not found in PATH.")
```

```python
# Check UV Package Manager
try:
    result = subprocess.run(["uv","--version"], capture_output=True, text=True, check=True)
    print("UV Package Manager is installed:", result.stdout.strip())
except subprocess.CalledProcessError:
    print("UV Package Manager is not installed or not found in PATH.")
```

```python
# Check Current Containers
import json 

try:
    result = subprocess.run(
        ["docker","compose","ps","--format","json"],
        capture_output=True,
        text=True,
        check=True
    )
    if result.returncode == 0 and result.stdout.strip():
        print("Current Docker Containers:")
        for line in result.stdout.strip().splitlines():
            if line.strip():
                try:
                    container_info = json.loads(line)
                    service = container_info.get("Service", "N/A")
                    state = container_info.get("State", "N/A")
                    ports = container_info.get("Ports", "N/A")
                    print(f"Service: {service}, State: {state}, Ports: {ports}")
                except json.JSONDecodeError:
                    print("Failed to decode JSON for a container entry.")
                    pass
    else:
        print("No running Docker containers found.")
except subprocess.CalledProcessError:
    print("Failed to retrieve Docker containers or Docker Compose is not installed.")
```
